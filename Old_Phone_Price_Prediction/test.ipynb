```python
# Install dependencies as needed:
# pip install kagglehub[pandas-datasets]
import kagglehub
from kagglehub import KaggleDatasetAdapter

# Set the path to the file you'd like to load
file_path = "used_phone_price_prediction_1M.csv"

# Load the latest version
df = kagglehub.load_dataset(
  KaggleDatasetAdapter.PANDAS,
  "sharmajicoder/used-phone-price-prediction-dataset",
  file_path,
  # Provide any additional arguments like 
  # sql_query or pandas_kwargs. See the 
  # documenation for more information:
  # https://github.com/Kaggle/kagglehub/blob/main/README.md#kaggledatasetadapterpandas
)

print("First 5 records:", df.head())
```

```python
# Display all available brands and their phone models

brand_model_list = (
    df[["brand", "model"]]
    .dropna()
    .drop_duplicates()
    .sort_values(["brand", "model"])
)

for brand, group in brand_model_list.groupby("brand"):
    print(f"\n{'=' * 40}")
    print(f"Brand: {brand}")
    print(f"{'=' * 40}")

    for model in group["model"]:
        print(f"  - {model}")
```

```python
brand_name = "Apple"

apple_models = sorted(
    df.loc[df["brand"] == brand_name, "model"]
    .dropna()
    .unique()
)

print(f"Models available for {brand_name}:")

for model in apple_models:
    print(model)
```

```python
df[["brand", "model"]].dropna().drop_duplicates().to_csv(
    "brand_model_catalog.csv",
    index=False
)
```
